```python
#1 Create an iterator from a list and print all elements using a while loop and next()
```

```python
numbers = [10, 20, 30, 40, 50]

it = iter(numbers)

while True:
    try:
        print(next(it))
    except StopIteration:
        break
```

```python
#2 Demonstrate what happens when next() is called more times than the number of elements
```

```python
numbers = [1, 2, 3]

it = iter(numbers)

print(next(it))
print(next(it))
print(next(it))

# One extra call
print(next(it))
```

```python
# 3 Iterate over an iterator using a for loop
```

```python
fruits = ["Apple", "Banana", "Mango"]

it = iter(fruits)

for item in it:
    print(item)
```

```python
#4 Create an iterator using range() and print values step by step using next()
```

```python
it = iter(range(1, 6))

print(next(it))
print(next(it))
print(next(it))
print(next(it))
print(next(it))
```

```python
#5 Manually traverse a list using iter() and next() and handle StopIteration
```

```python
names = ["Ankit", "Rahul", "Amit"]

it = iter(names)

try:
    while True:
        print(next(it))
except StopIteration:
    print("Iteration Completed")
```

```python
# Generators
```

```python
#Generator function using yield that returns multiple values and print them using next()
```

```python
def numbers():
    yield 10
    yield 20
    yield 30

gen = numbers()

print(next(gen))
print(next(gen))
print(next(gen))
```

```python
#Iterate over a generator using a for loop
```

```python
def colors():
    yield "Red"
    yield "Green"
    yield "Blue"

for color in colors():
    print(color)
```

```python
#Generator expression to generate square values and print first few elements using next()
```

```python
square = (x*x for x in range(1, 6))

print(next(square))
print(next(square))
print(next(square))
print(next(square))
print(next(square))
```

```python
#Generator function to generate Fibonacci series and print the output as a list
```

```python
def fibonacci(n):
    a, b = 0, 1

    for i in range(n):
        yield a
        a, b = b, a + b

result = list(fibonacci(10))

print(result)
```

```python
#Demonstrate generator methods like next() and send()
```

```python
def demo():
    print("Generator Started")

    x = yield "First Yield"
    print("Received:", x)

    y = yield "Second Yield"
    print("Received:", y)

gen = demo()

print(next(gen))          # Starts generator

print(gen.send("Hello"))  # Sends value to x

try:
    print(gen.send("Python"))  # Sends value to y
except StopIteration:
    print("Generator Finished")
```
