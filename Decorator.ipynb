```python
#1 Create a simple decorator that adds extra print statements before and after a function execution
```

```python
def my_decorator(func):
    def wrapper():
        print("Before function execution")
        func()
        print("After function execution")
    return wrapper


@my_decorator
def greet():
    print("Hello!")

greet()
```

```python
#2 Create a decorator that wraps a function and prints a custom message before calling the original function
```

```python
def message_decorator(func):
    def wrapper():
        print("Welcome! Function is about to execute.")
        func()
    return wrapper


@message_decorator
def display():
    print("This is the original function.")

display()
```

```python
#3 Create a decorator that takes a function with two arguments and logs the function name along with its parameters
```

```python
def log_decorator(func):
    def wrapper(a, b):
        print(f"Function Name: {func.__name__}")
        print(f"Arguments: {a}, {b}")
        return func(a, b)
    return wrapper


@log_decorator
def add(a, b):
    return a + b

print(add(10, 20))
```

```python
#4 Create a decorator to measure execution time of a function that takes no arguments
```

```python
import time

def timer(func):
    def wrapper():
        start = time.time()
        func()
        end = time.time()
        print("Execution Time:", end - start, "seconds")
    return wrapper


@timer
def test():
    time.sleep(2)
    print("Function Executed")

test()
```

```python
#5 Create a decorator that checks if the input user is "admin" before allowing function execution
```

```python
def admin_only(func):
    def wrapper(user):
        if user == "admin":
            return func(user)
        else:
            print("Access Denied!")
    return wrapper


@admin_only
def dashboard(user):
    print("Welcome Admin!")

dashboard("admin")
dashboard("guest")
```

```python
#6 Create a decorator that caches results of a function with one integer argument to avoid repeated calculations
```

```python
def cache_decorator(func):
    cache = {}

    def wrapper(n):
        if n in cache:
            print("Returning Cached Result")
            return cache[n]

        result = func(n)
        cache[n] = result
        return result

    return wrapper


@cache_decorator
def square(n):
    print("Calculating...")
    return n * n

print(square(5))
print(square(5))
```

```python
# Create a decorator that works with functions having arguments using *args and **kwargs and prints messages before and after execution
```

```python
def decorator(func):
    def wrapper(*args, **kwargs):
        print("Before Execution")
        result = func(*args, **kwargs)
        print("After Execution")
        return result
    return wrapper


@decorator
def multiply(a, b):
    print("Multiplication:", a * b)

multiply(4, 5)
```

```python
#8 Create a decorator that modifies the return value of a function by doubling the result
```

```python
def double_result(func):
    def wrapper():
        result = func()
        return result * 2
    return wrapper


@double_result
def get_number():
    return 25

print(get_number())
```
